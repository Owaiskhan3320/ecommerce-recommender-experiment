# 04 — Recommender Experiment Design

**Question:** what would a fair production experiment look like for the co-visitation recommender?

This is an experiment plan, not a result. The real clickstream data supplies a historical planning proxy only; it does not identify causal impact.

```python
from pathlib import Path

import pandas as pd

project_root = Path.cwd()
if project_root.name == 'notebooks':
    project_root = project_root.parent
reports_directory = project_root / 'reports'

sample_size_scenarios = pd.read_csv(reports_directory / 'phase_4_sample_size_scenarios.csv')
sample_size_scenarios
```

## Recommended setup

- Compare co-visitation rankings with a global-popularity ranking in the **same** recommendation slot.
- Assign an eligible visitor persistently to one variant; analyze their first qualified assignment, including failed renders.
- Use purchase conversion after exposure as the primary metric, then evaluate click-through and add-to-cart as secondary diagnostics.
- Hold the experiment until both the pre-calculated sample size and a 14-day minimum have been reached.
- Report an intent-to-treat conversion difference and a 95% confidence interval; do not use offline recall as evidence of lift.

## What must be logged

At minimum: experiment ID, variant, persistent visitor/device ID, session ID, assignment timestamp, exposure timestamp, anchor item, shown candidate items and ranks, render outcome, candidate click, cart event, and transaction. Assignment must be logged before rendering so every assigned eligible visitor, including render failures, remains in the intent-to-treat analysis.

## Decision boundary

Ship only when the primary conversion confidence interval is fully positive, guardrails pass, and the result meets the chosen business threshold. Otherwise, keep the baseline or investigate the module; never claim causality from the historical clickstream alone.
