## Offline metrics

`Recall@K` is the percentage of eligible holdout purchase examples where the actual purchased item appears in the top K candidates. Candidate coverage is reported separately because the co-visitation model cannot recommend for every rare or new anchor item. The popularity baseline recommends the most common training-session items to every example.

## Honest interpretation

This is a pragmatic baseline for an anonymous-session dataset, not a deployed personalization system. A better score than popularity shows that co-visitation adds retrieval signal in historical behavior; it does not prove product impact. Before product use, inspect item quality manually, define display rules, and run a randomized experiment with a conversion metric.
