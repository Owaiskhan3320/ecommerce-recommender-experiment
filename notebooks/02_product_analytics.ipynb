# 02 — Product Analytics

## Question
Where is the biggest observed drop-off in the session journey, and what product problem should be investigated next?

This notebook reads the reproducible outputs generated by `src/run_product_analysis.py`. The funnel is strict: a cart must follow a view, and a transaction must follow that cart.


```python
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

project_root = Path.cwd().resolve()
if project_root.name == 'notebooks':
    project_root = project_root.parent

reports_directory = project_root / 'reports'
figures_directory = reports_directory / 'figures'
```

## Strict session funnel
A session enters the last stage only after an observed product view, add-to-cart, and transaction in that order.


```python
session_funnel = pd.read_csv(reports_directory / 'phase_2_session_funnel.csv')
session_funnel

display(Image(filename=figures_directory / 'phase_2_strict_session_funnel.png'))
```

## First versus later observed sessions
This is descriptive only. A later observed session is not evidence that returning causes a transaction.


```python
first_vs_later = pd.read_csv(reports_directory / 'phase_2_first_vs_later_sessions.csv')
first_vs_later

display(Image(filename=figures_directory / 'phase_2_first_vs_later_sessions.png'))
```

## Session duration
Session duration is another descriptive indicator of engagement. It must not be interpreted as a causal lever.


```python
session_duration = pd.read_csv(reports_directory / 'phase_2_session_duration.csv')
session_duration

display(Image(filename=figures_directory / 'phase_2_session_duration.png'))
```

## Recommendation
The view-to-cart transition is the largest observed drop-off. The next project phase evaluates whether a simple item-to-item recommender can improve product retrieval quality offline. That evaluation does not prove an online conversion effect; a controlled experiment would be required.
